```python
health_df = health_df.dropna()
```

```python
health_df["Date of Admission"] = pd.to_datetime(health_df["Date of Admission"], utc=True)
health_df["Date of Admission"].dtype
```

```python
health_df["Discharge Date"] = pd.to_datetime(health_df["Discharge Date"], utc=True)
health_df["Discharge Date"].dtype
```

```python
health_df["LengthOfStay_Days"] = (health_df["Discharge Date"] - health_df["Date of Admission"]).dt.days
health_df
```

```python
cost_summary = health_df.groupby(["Medical Condition", "Insurance Provider"])["Billing Amount"].sum().unstack()
cost_summary
```

```python
cost_summary.plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6.5),
    colormap="tab20"
)
plt.xlabel("Medical Condition")
plt.ylabel("Total Billing Amount")
plt.title("Total Billing Amount by Medical Condition and Insurance Provider")
plt.xticks(rotation=40)
plt.legend(title="Insurance Provider")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11, 6.5))
sns.violinplot(
    x="Medical Condition",
    y="LengthOfStay_Days",
    data=health_df,
    palette="Set2"
)
plt.title("Distribution of Hospital Stay Length by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Length of Stay (days)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
health_df["AdmissionMonth"] = health_df["Date of Admission"].dt.to_period("M")
health_df["AdmissionMonth"]
```

```python
admissions_trend = health_df.groupby("AdmissionMonth").size()
admissions_trend
```

```python
plt.figure(figsize=(11, 6.5))
admissions_trend.plot(kind="line", marker="s", color="darkorange")
plt.title("Monthly Patient Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation=40)
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
corr_matrix = health_df[["Age", "LengthOfStay_Days", "Billing Amount"]].corr()
corr_matrix
```

```python
plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, cmap="viridis", fmt=".2f")
plt.title("Correlation Matrix: Age, Length of Stay, Billing Amount")
plt.tight_layout()
plt.show()
```

# Executive Summary — Healthcare Dataset EDA (Week 6)

## Objective

This notebook explores patient-level hospital records to understand cost drivers, stay duration, and admission patterns. The goal is to surface operational and financial trends across medical conditions, insurers, and time that can support cost control, capacity planning, and later predictive work.

## Approach

The workflow follows four stages:

1. **Cleaning and Preparation** — The raw dataset was loaded and checked for missing values, which were dropped to keep the analysis reliable. Admission and discharge timestamps were parsed into proper datetime objects so time-based math could be performed.
2. **Derived Features** — Two new fields were engineered that weren't present in the source data: **Length of Stay** (discharge date minus admission date, in days) and **Admission Month** (used to group records for trend analysis).
3. **Cost Breakdown** — Billing totals were aggregated by medical condition and insurance provider to see which combinations account for the largest spend.
4. **Clinical & Time-Based Patterns** — Stay-length distributions were compared across conditions, admissions were tracked on a monthly basis, and the statistical relationship between age, stay length, and billing was examined.

## Chart Guide

- **Stacked Bar (Billing by Condition & Insurer):** Total billing per medical condition, broken into segments by insurer — highlights which condition/insurer pairs drive the most spend.
- **Violin Plot (Stay Length Distribution):** Full distribution of stay lengths per condition, showing spread and clustering rather than just an average.
- **Line Chart (Monthly Admissions):** Admission counts per month, useful for spotting trends or seasonality.
- **Heatmap (Correlation Matrix):** Pairwise correlation strength between Age, Length of Stay, and Billing Amount, on a -1 to +1 scale.

## Key Questions Answered

| Analysis | Question |
|---|---|
| Billing by Condition & Insurer | Which condition/insurer combinations generate the highest total spend? |
| Stay Length Distribution | Which conditions tend toward longer or more variable stays? |
| Monthly Admissions Trend | Is patient volume trending up, down, or seasonally? |
| Correlation Analysis | Do age or stay length relate to higher billing amounts? |
