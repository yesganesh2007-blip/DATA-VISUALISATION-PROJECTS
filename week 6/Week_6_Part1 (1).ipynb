```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
health_df = health_df.dropna()
```

```python
health_df["Date of Admission"] = pd.to_datetime(health_df["Date of Admission"], utc=True)
health_df["Date of Admission"].dtype
```

```python
health_df["Discharge Date"] = pd.to_datetime(health_df["Discharge Date"], utc=True)
health_df["Discharge Date"].dtype
```

```python
health_df["LengthOfStay_Days"] = (health_df["Discharge Date"] - health_df["Date of Admission"]).dt.days
health_df
```

```python
cost_summary = health_df.groupby(["Medical Condition", "Insurance Provider"])["Billing Amount"].sum().unstack()
cost_summary
```

```python
cost_summary.plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6.5),
    colormap="tab20"
)
plt.xlabel("Medical Condition")
plt.ylabel("Total Billing Amount")
plt.title("Total Billing Amount by Medical Condition and Insurance Provider")
plt.xticks(rotation=40)
plt.legend(title="Insurance Provider")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11, 6.5))
sns.violinplot(
    x="Medical Condition",
    y="LengthOfStay_Days",
    data=health_df,
    palette="Set2"
)
plt.title("Distribution of Hospital Stay Length by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Length of Stay (days)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
